## Parse sample HTML

```python
from bs4 import BeautifulSoup
import pandas as pd

html = """
<html>
  <head><title>Example page</title></head>
  <body>
    <h1>Data Science Resources</h1>
    <p class="description">A page for practicing HTML parsing.</p>
    <ul>
      <li><a href="/intro">Introduction</a></li>
      <li><a href="/pandas">Pandas</a></li>
      <li><a href="/visualization">Visualization</a></li>
    </ul>
  </body>
</html>
"""

soup = BeautifulSoup(html, "html.parser")
print("Title:", soup.title.get_text(strip=True))
print("Heading:", soup.find("h1").get_text(strip=True))
print("Description:", soup.select_one(".description").get_text(strip=True))
```

## Extract links

```python
links = [
    {"text": tag.get_text(" ", strip=True), "href": tag["href"]}
    for tag in soup.select("a[href]")
]
pd.DataFrame(links)
```
