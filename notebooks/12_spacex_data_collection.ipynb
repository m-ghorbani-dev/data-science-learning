## Learning objectives
Inspect the prepared SpaceX launch dataset and identify useful fields.

The CSV path is relative to the `notebooks/` directory.

```python
import pandas as pd

file_path = "../data/raw/dataset_falcon9.csv"
df = pd.read_csv(file_path)
print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

## Inspect the dataset

```python
display(df.head())
df.info()
print("Columns:")
print(df.columns.tolist())
```

## Missing values and target distribution

```python
display(df.isna().sum().sort_values(ascending=False).to_frame("missing_count"))
if "Class" in df.columns:
    display(df["Class"].value_counts(dropna=False).to_frame("count"))
    display(df["Class"].value_counts(normalize=True, dropna=False).to_frame("proportion"))
```

## Review candidate columns

```python
candidate_columns = ["Date", "BoosterVersion", "PayloadMass", "Orbit", "LaunchSite", "Flights", "GridFins", "Reused", "Legs", "LandingPad", "Block", "Class"]
available_columns = [column for column in candidate_columns if column in df.columns]
display(df[available_columns].head())
```
